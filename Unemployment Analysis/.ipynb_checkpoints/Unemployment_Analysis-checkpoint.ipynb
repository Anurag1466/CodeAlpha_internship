# Unemployment Analysis — Impact of Covid-19

## 1. Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
pd.set_option('display.max_columns', None)
```

## 2. Load Data

```python
df1_raw = pd.read_csv('Unemployment_in_India.csv')
df2_raw = pd.read_csv('Unemployment_Rate_upto_11_2020.csv')
df1_raw.shape, df2_raw.shape
```

## 4. Data Cleaning

```python
df1 = df1_raw.copy()
df2 = df2_raw.copy()

df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

df1 = df1.dropna(how='all').copy()

for df in (df1, df2):
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()

df1.shape
```

```python
rename_map = {
    'Estimated Unemployment Rate (%)': 'unemployment_rate',
    'Estimated Employed': 'employed',
    'Estimated Labour Participation Rate (%)': 'labour_participation_rate',
    'Region': 'state',
    'Date': 'date',
    'Frequency': 'frequency',
    'Area': 'area',
}

df1 = df1.rename(columns=rename_map)
df2 = df2.rename(columns={**rename_map, 'Region.1': 'zone'})

df1['date'] = pd.to_datetime(df1['date'], format='%d-%m-%Y')
df2['date'] = pd.to_datetime(df2['date'], format='%d-%m-%Y')

df1['date'].min(), df1['date'].max(), df2['date'].min(), df2['date'].max()
```

```python
df1['unemployment_rate'].min(), df1['unemployment_rate'].max(), df2['unemployment_rate'].min(), df2['unemployment_rate'].max()
```

```python
df1.duplicated(subset=['state','date','area']).sum(), df2.duplicated(subset=['state','date']).sum()
```

## 5. Building Analysis Datasets

```python
df_area = df1[['state','date','area','unemployment_rate','employed','labour_participation_rate']].copy()
df_area = df_area.sort_values(['state','area','date']).reset_index(drop=True)
df_area.head()
```

```python
df1_state = (df1.groupby(['state','date'], as_index=False)
             .agg(unemployment_rate=('unemployment_rate','mean'),
                  employed=('employed','sum'),
                  labour_participation_rate=('labour_participation_rate','mean')))
df1_state['source'] = 'file1_avg'

df2_state = df2[['state','date','unemployment_rate','employed','labour_participation_rate','zone','longitude','latitude']].copy()
df2_state['source'] = 'file2'

cutoff = df1_state['date'].max()
df2_tail = df2_state[df2_state['date'] > cutoff].copy()

df_state = pd.concat([df1_state, df2_tail], ignore_index=True, sort=False)
df_state = df_state.sort_values(['state','date']).reset_index(drop=True)
df_state.shape
```

```python
def covid_period(date):
    if date < pd.Timestamp('2020-03-01'):
        return 'Pre-Covid'
    elif date <= pd.Timestamp('2020-06-30'):
        return 'Covid Peak (Lockdown)'
    else:
        return 'Recovery'

df_state['covid_period'] = df_state['date'].apply(covid_period)
df_state['year'] = df_state['date'].dt.year
df_state['month'] = df_state['date'].dt.month
df_state['month_name'] = df_state['date'].dt.strftime('%b')
df_state.tail()
```

## 6. Exploratory Data Analysis

### National Trend

```python
national_trend = df_state.groupby('date')['unemployment_rate'].mean().reset_index()

plt.figure(figsize=(13,5))
plt.plot(national_trend['date'], national_trend['unemployment_rate'], marker='o', color='crimson')
plt.axvspan(pd.Timestamp('2020-03-01'), pd.Timestamp('2020-06-30'), color='orange', alpha=0.2, label='Covid Lockdown Period')
plt.title('National Average Unemployment Rate Over Time')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Unemployment by State

```python
state_avg = df_state.groupby('state')['unemployment_rate'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,8))
sns.barplot(x=state_avg.values, y=state_avg.index, palette='rocket')
plt.title('Average Unemployment Rate by State')
plt.xlabel('Avg Unemployment Rate (%)')
plt.tight_layout()
plt.show()
```

## 7. Rural vs Urban

```python
area_trend = df_area.groupby(['date','area'])['unemployment_rate'].mean().reset_index()

plt.figure(figsize=(12,5))
sns.lineplot(data=area_trend, x='date', y='unemployment_rate', hue='area', marker='o')
plt.title('Rural vs Urban Unemployment Rate Over Time')
plt.ylabel('Unemployment Rate (%)')
plt.xlabel('Date')
plt.tight_layout()
plt.show()
```

```python
df_area.groupby('area')['unemployment_rate'].agg(['mean','max','std'])
```

## 8. Covid-19 Impact

```python
period_summary = df_state.groupby('covid_period')['unemployment_rate'].agg(['mean','max','min','std']).reindex(
    ['Pre-Covid','Covid Peak (Lockdown)','Recovery'])
period_summary
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df_state, x='covid_period', y='unemployment_rate',
            order=['Pre-Covid','Covid Peak (Lockdown)','Recovery'], palette='Set2')
plt.title('Unemployment Rate by Covid Period')
plt.ylabel('Unemployment Rate (%)')
plt.xlabel('')
plt.tight_layout()
plt.show()
```

```python
pre_covid_avg = df_state[df_state['covid_period']=='Pre-Covid']['unemployment_rate'].mean()
peak_row = national_trend.loc[national_trend['unemployment_rate'].idxmax()]
latest = national_trend.iloc[-1]

pre_covid_avg, peak_row['date'], peak_row['unemployment_rate'], latest['date'], latest['unemployment_rate']
```

```python
peak_month = peak_row['date']
peak_by_state = df_state[df_state['date']==peak_month].sort_values('unemployment_rate', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(14,6))
sns.barplot(data=peak_by_state.head(10), x='unemployment_rate', y='state', ax=axes[0], palette='Reds_r')
axes[0].set_title(f'Hardest Hit States — {peak_month.strftime("%B %Y")}')
axes[0].set_xlabel('Unemployment Rate (%)')

sns.barplot(data=peak_by_state.tail(10), x='unemployment_rate', y='state', ax=axes[1], palette='Greens_r')
axes[1].set_title(f'Most Resilient States — {peak_month.strftime("%B %Y")}')
axes[1].set_xlabel('Unemployment Rate (%)')
plt.tight_layout()
plt.show()
```

## 9. Seasonality

```python
month_order = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']

plt.figure(figsize=(12,5))
sns.boxplot(data=df_state, x='month_name', y='unemployment_rate', order=month_order, palette='coolwarm')
plt.title('Unemployment Rate by Month')
plt.ylabel('Unemployment Rate (%)')
plt.xlabel('Month')
plt.tight_layout()
plt.show()
```

```python
ts = national_trend.set_index('date')['unemployment_rate'].asfreq('ME')
ts = ts.interpolate()

if len(ts) >= 24:
    decomposition = seasonal_decompose(ts, model='additive', period=12)
    fig = decomposition.plot()
    fig.set_size_inches(11, 8)
    plt.tight_layout()
    plt.show()
else:
    print(f"Only {len(ts)} months available — need 24+ for reliable decomposition")
```

## 10. Labour Participation vs Unemployment

```python
fig, ax1 = plt.subplots(figsize=(13,5))
ax2 = ax1.twinx()

ax1.plot(national_trend['date'], national_trend['unemployment_rate'], color='crimson', marker='o', label='Unemployment Rate')
lpr_trend = df_state.groupby('date')['labour_participation_rate'].mean().reset_index()
ax2.plot(lpr_trend['date'], lpr_trend['labour_participation_rate'], color='steelblue', marker='s', label='Labour Participation Rate')

ax1.set_ylabel('Unemployment Rate (%)', color='crimson')
ax2.set_ylabel('Labour Participation Rate (%)', color='steelblue')
ax1.set_xlabel('Date')
plt.title('Unemployment Rate vs Labour Participation Rate')
fig.tight_layout()
plt.show()
```

## 11. Key Insights

- Unemployment spiked sharply during the Covid lockdown, then recovered by Oct 2020
- Rural and urban areas were affected differently
- Some states were hit far harder than others at the peak
- Recurring monthly patterns suggest structural seasonality independent of Covid
- Labour participation rate trends alongside unemployment, indicating discouraged-worker effects
